<h2>A quantitative model of monopoly</h2>

With __R__, we can create a simple numerical model of monopoly that draws on [Example 8.3](https://prod.reader-ui.prod.mheducation.com/epub/sn_4a367/data-uuid-198e221c895d456295ea32ac378b2c81) in Frank, Bernanke, Antonovics, and Heffetz, <i>Principles of Economics</i>. We can get our hands dirty with demand, marginal revenue $MR$, marginal cost $MC$, and profit maximization.


<b>Learning objectives:</b>

1. Profit maximization occurs at a quantity $Q_m$ where $MR = MC$ 
2. The monopolist sets the price $P_m$ using the level of demand at $Q_m$
3. Graphical or numerical solution techniques are useful tools in the economist's toolkit
4. Fixed costs absorbs monopoly profit but do not affect profit-maximizing output
5. Shifts in $MC$, like shifts in demand, change things

```r
# Let's install and load up the handy graphing library ggplot
install.packages("ggplot2")
library(ggplot2)
```

<h3>Assumptions</h3>

Let's follow Example 8.3 in the textbook, which yields Figure 8.7 shown below.

<img src="images/figure8.7.png" width="400" />

In this example, we can see that demand, MR, and MC are all linear, with these functional forms:

$$
P = D(Q) = 6 - \frac{1}{4}  Q
$$
$$
MR = 6 - \frac{1}{2} Q
$$
$$
MC = \frac{1}{4} Q
$$

To begin, we will <i>assume that fixed costs are zero</i>. Fixed costs do not affect marginal cost in any event, because they are fixed. But profit will depend on fixed costs. 

<h3>Using <b>R</b> for numerical simulation</h3>

In __R__, we can use some simple expressions to set up a numerical model of a market with prices and quantities. Here is Example 8.3, where quantity runs in integers from 0 to 24 units. The code below creates this vector for $Q$:

```r
Q = 0:24
Q
```

__R__ understands the difference between a vector and a scalar, so a function of a scalar times a vector is a vector. That's a fancy way of saying, "writing the algebra produces the outcome we'd like to see." Below is a vector of quantities demanded that we derive using the demand relationship between price $P$ and quantity $Q$:

$$
P = D(Q) = 6 - \frac{1}{4}  Q ,
$$

As shown in Figure 8.7 above, this is a downward sloping linear demand curve with a $y$-intercept equal to $6$, and a slope equal to negative one quarter or $-0.25$.

```r
D = 6 - 0.25 * Q
D
```

In __R__, a lot of helpful functions require us to place data like these two vectors into a <b>data frame</b>, which is a matrix that has columns labeled with names. Here is code that places these two vectors of data into a data frame and labels the columns with "Quantity" and "Demand."

```r
monopdata = data.frame(Quantity = Q, Demand = D)
```

With our market data in a data frame, we can get back to business with creating new columns that are functions of the Quantity vector. Here is <b>marginal cost (MC)</b>, given by the mathematical relationship from above:
$$
MC = \frac{1}{4} Q
$$

```r
monopdata$MC = 0.25 * monopdata$Quantity
```

<b>Marginal revenue (MR)</b> is the weird one. When demand is linear, it turns out that marginal revenue is also linear, with a slope that is twice as steep. 

$$
MR = 6 - \frac{1}{2} Q
$$

This derives from calculus, but you do not need to master calculus. You should internalize this rule, that the MR curve is twice as steep as demand with the same $y$-intercept.

```r
monopdata$MR = 6 - 0.5 * monopdata$Quantity
```

<hr>

<b>Total cost</b> is the area under marginal cost <i>plus any fixed costs</i>. For now, we have assumed that fixed costs are zero, but later we can relax that assumption. The area under $MC$ is the integral of $MC$ from $0$ to $Q$.

$$
TC = \int_0^Q MC \ dq = \int_0^Q \frac{1}{4} q \ dq = 
\left[ \frac{1}{8} q^2\right]_0^Q
= \frac{1}{8} Q^2
$$

```r
monopdata$TC = (1/8) * monopdata$Quantity^2
```

<b>Total revenue</b> is the product of the price, which is given by the demand curve, and the quantity:

$$
TR = P \times Q = D(Q) \times Q
$$

```r
monopdata$TR = monopdata$Demand * monopdata$Quantity
```

<b>Profit</b> is the difference between total revenue and total cost. When fixed costs are zero, they do not appear:

$$
\Pi = TR - TC = D(Q) \times Q - \frac{1}{8}Q^2
$$

```r
monopdata$Pi = monopdata$TR - monopdata$TC
```

<h3>Graphical solutions</h3>

We have generated all the functions, and now we are ready to look at them in graphs. Here is a mess of `ggplot` syntax. I find it pretty confusing, and my suggestion is to copy and paste and then alter it. Pro tips: those "+" signs are important, and setting the $x$ and $y$ scales and limits are important for good visualizations.

<h3>Replication of Figure 8.7</h3>

```r
scatterplot2 <- ggplot(monopdata, aes(x = Quantity)) +
  geom_line(aes(y = Demand, color = "Demand"), linewidth = 1, color = "blue") + 
  geom_line(aes(y = MC, color = "MC"), linewidth = 1, color = "red") +
  geom_line(aes(y = MR, color = "MR"), linewidth = 1, color = "darkgreen") +
  labs(
    title = "Figure 8.7",
    x = "Quantity",
    y = "Price"
  ) +
  theme_minimal() +
  scale_x_continuous(breaks = seq(from = 0, to = 25, by = 1)) +
  scale_y_continuous(breaks = seq(from = 0, to = 6, by = 1)) +
  coord_cartesian(
    xlim = c(0, 25),
    ylim = c(0, 6)
  )
print(scatterplot2)
```

Like we saw in Figure 8.7, the graphic above reveals that $MR = MC$ occurs when $Q = 8$. That is the quantity the monopolist will set, and we can read the market price off the demand curve: $P = 4$.

Rather than take my word for it, you could also graph <b>total revenue</b>, <b>total cost</b>, and <b>profit</b> using our numerical model. Then the objective would be to look for the maximum profit. Here is code that plots these three functions using the same $x$-axis scale but updated values for `scale_y_continuous` and `ylim`:

```r
scatterplot3 <- ggplot(monopdata, aes(x = Quantity)) +
  geom_line(aes(y = TR, color = "Total revenue"), linewidth = 1) + 
  geom_line(aes(y = TC, color = "Total cost"), linewidth = 1) + 
  geom_line(aes(y = Pi, color = "Profit"), linewidth = 1) + 
  labs(
    title = "Profit, cost, and revenue",
    x = "Quantity",
    y = "Profit, cost, and revenue"
  ) +
  theme_minimal() +
#  theme_minimal(base_size = 20) +
#  theme_gray(base_size = 20) +
  scale_x_continuous(breaks = seq(from = 0, to = 25, by = 1)) +
  scale_y_continuous(breaks = seq(from = 0, to = 80, by = 10)) +
  coord_cartesian(
    xlim = c(0, 25),
    ylim = c(0, 60)
  )
print(scatterplot3)
```

Can you visually identify the quantity where profit (the red curve) is maximized? What is it? Is that the same quantity that we saw above when setting $MR = MC$?

<h3>Extension 1: Fixed costs</h3>

Suppose the monopolist faced large fixed costs of operation. This could happen if it were a public utility like [Pacific Gas and Electric (PG&E)](https://www.pge.com/), with very expensive infrastructure that is required to supply electricity and natural gas.

Imagine that fixed costs per period are $FC = 30$.

Does $FC = 30$ change Figure 8.7, which shows marginal cost $MC$?

Does $FC = 30$ change profit $\Pi$?

Use __R__ to model profit, total cost, and total revenue when $FC = 30$. We create new columns TC2 and Pi2 that reflect $FC$ in the code below:

```r
monopdata$TC1 = monopdata$TC + 30
monopdata$Pi1 = monopdata$TR - monopdata$TC1
```

```r
scatterplot3 <- ggplot(monopdata, aes(x = Quantity)) +
  geom_line(aes(y = TR, color = "Total revenue"), linewidth = 1) + 
  geom_line(aes(y = TC1, color = "Total cost"), linewidth = 1) + 
  geom_line(aes(y = Pi1, color = "Profit"), linewidth = 1) + 
  labs(
    title = "Profit, cost, and revenue with fixed costs",
    x = "Quantity",
    y = "Profit, cost, and revenue"
  ) +
  theme_minimal() +
#  theme_minimal(base_size = 20) +
#  theme_gray(base_size = 20) +
  scale_x_continuous(breaks = seq(from = 0, to = 25, by = 1)) +
  scale_y_continuous(breaks = seq(from = -30, to = 50, by = 10)) +
  coord_cartesian(
    xlim = c(0, 25),
    ylim = c(-30, 60)
  )
print(scatterplot3)
```

Discuss. Where will the monopolist operate? Will they try to do anything else?

<h3>Extension 2: A shift in $MC$</h3>

Suppose $MC$ rose by $1.5$. Can you model that numerically? (Does it help to graph it by hand first, or run the algebra?)
$$
MC = \frac{1}{4} Q + 1.5
$$

```r
monopdata$MC2 = 0.25 * monopdata$Q + 1.5
```

```r
scatterplot2 <- ggplot(monopdata, aes(x = Quantity)) +
  geom_line(aes(y = Demand, color = "Demand"), linewidth = 1, color = "blue") + 
  geom_line(aes(y = MC, color = "MC"), linewidth = 1, color = "red") +
  geom_line(aes(y = MC2, color = "MC"), linewidth = 1, color = "darkred") +
  geom_line(aes(y = MR, color = "MR"), linewidth = 1, color = "darkgreen") +
  labs(
    title = "Figure 8.7",
    x = "Quantity",
    y = "Price"
  ) +
  theme_minimal() +
  scale_x_continuous(breaks = seq(from = 0, to = 25, by = 1)) +
  scale_y_continuous(breaks = seq(from = 0, to = 6, by = 1)) +
  coord_cartesian(
    xlim = c(0, 25),
    ylim = c(0, 6)
  )
print(scatterplot2)
```

Recall that if we want to calculate profit, a shift in marginal cost like this also affects total cost:
$$
TC2 = \int_0^Q (MC + 1.5)\ dq = \int_0^Q \left(\frac{1}{4} q + 1.5 \right) \ dq = 
\left[ \frac{1}{8} q^2 + 1.5 q\right]_0^Q
= \frac{1}{8} Q^2 + 1.5 \ Q
$$

```r
monopdata$TC2 = (1/8) * monopdata$Quantity^2 + 1.5 * monopdata$Quantity
monopdata$Pi2 = monopdata$TR - monopdata$TC2
```

```r
scatterplot3 <- ggplot(monopdata, aes(x = Quantity)) +
  geom_line(aes(y = TR, color = "Total revenue"), linewidth = 1) + 
  geom_line(aes(y = TC2, color = "Total cost"), linewidth = 1) + 
  geom_line(aes(y = Pi2, color = "Profit"), linewidth = 1) + 
  labs(
    title = "Profit, cost, and revenue with upward shifted MC",
    x = "Quantity",
    y = "Profit, cost, and revenue"
  ) +
  theme_minimal() +
#  theme_minimal(base_size = 20) +
#  theme_gray(base_size = 20) +
  scale_x_continuous(breaks = seq(from = 0, to = 25, by = 1)) +
  scale_y_continuous(breaks = seq(from = 0, to = 80, by = 10)) +
  coord_cartesian(
    xlim = c(0, 25),
    ylim = c(0, 60)
  )
print(scatterplot3)
```

Discuss. What is the new quantity? Is it the same as what you saw implied above in the $MR = MC$ graph?
