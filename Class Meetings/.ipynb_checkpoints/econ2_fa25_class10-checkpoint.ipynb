<h2>A quantitative model of monopoly</h2>

With __R__, we can create a simple numerical model of monopoly that draws on Example 8.3 in Frank, Bernanke, Antonovics, and Heffetz, <i>Principles of Economics</i>. We can get our hands dirty with demand, marginal revenue ($MR$), marginal cost ($MC$), and profit maximization.

Learning objectives:

1. Profit maximization occurs at a quantity $Q_m$ where $MR = MC$ 
2. The monopolist sets the price $P_m$ using the level of demand at $Q_m$

```r
# Let's install and load up the handy graphing library ggplot
install.packages("ggplot2")
library(ggplot2)
```

In __R__, we can use some simple expressions to set up a numerical model of a market with prices and quantities. Here is Example 8.3, where quantity runs in integers from 0 to 24 units. The code below creates a vector for $Q$:

```r
Q = 0:24
Q
```

__R__ understands the difference between a vector and a scalar, so a function of a scalar times a vector is a vector. That's a fancy way of saying, "writing the algebra produces the outcome we'd like to see." Below is a vector of demand using the following demand relationship between price $P$ and quantity $Q$:

$$
P = D(Q) = 6 - 0.25 \cdot Q ,
$$

which is a downward sloping linear demand curve with $y$-intercept equal to $6$ and slope $-0.25$ or negative one quarter.

```r
D = 6 - 0.25 * Q
D
```

In __R__, a lot of helpful functions require us to place data like these two vectors into a <b>data frame</b>, which is a matrix that has columns labeled with names. Here is code that places the data into a data frame and labels the columns with "Quantity" and "Demand."

```r
monopdata <- data.frame(Quantity = Q, Demand = D)
```

With our market data in a data frame, we can get back to business with creating new columns that are functions of Quantity.

```r
monopdata$MC = 0.25 * monopdata$Quantity
```

```r
monopdata$MR = 6 - 0.5 * monopdata$Quantity
```

```r
monopdata$TC = (1/8) * monopdata$Quantity^2
```

```r
monopdata$TR = monopdata$Demand * monopdata$Quantity
```

```r
monopdata$Pi = monopdata$TR - monopdata$TC
```

```r
scatterplot1 <- ggplot(monopdata, aes(x = Quantity)) +
  geom_line(aes(y = Demand, color = "Demand"), linewidth = 1, color = "blue") + 
  geom_line(aes(y = MC, color = "MC"), linewidth = 1, color = "red") +
  labs(
    title = "Title",
    x = "Quantity",
    y = "Price"
  ) +
  theme_minimal() +
#  theme_minimal(base_size = 20) +
#  theme_gray(base_size = 20) +
  scale_x_continuous(breaks = seq(from = 0, to = 25, by = 1)) +
  scale_y_continuous(breaks = seq(from = 0, to = 6, by = 1)) +
  coord_cartesian(
    xlim = c(0, 25),
    ylim = c(0, 6)
  )
print(scatterplot1)
```

```r
scatterplot2 <- ggplot(monopdata, aes(x = Quantity)) +
  geom_line(aes(y = Demand, color = "Demand"), linewidth = 1, color = "blue") + 
  geom_line(aes(y = MC, color = "MC"), linewidth = 1, color = "red") +
  geom_line(aes(y = MR, color = "MR"), linewidth = 1, color = "green") +
  labs(
    title = "Title",
    x = "Quantity",
    y = "Price"
  ) +
  theme_minimal() +
#  theme_minimal(base_size = 20) +
#  theme_gray(base_size = 20) +
  scale_x_continuous(breaks = seq(from = 0, to = 25, by = 1)) +
  scale_y_continuous(breaks = seq(from = 0, to = 6, by = 1)) +
  coord_cartesian(
    xlim = c(0, 25),
    ylim = c(0, 6)
  )
print(scatterplot2)
```

```r
scatterplot3 <- ggplot(monopdata, aes(x = Quantity)) +
  geom_line(aes(y = TR, color = "Total revenues"), linewidth = 1) + 
  geom_line(aes(y = TC, color = "Total cost"), linewidth = 1) + 
  geom_line(aes(y = Pi, color = "Profits"), linewidth = 1) + 
  labs(
    title = "Title",
    x = "Quantity",
    y = "Price"
  ) +
  theme_minimal() +
#  theme_minimal(base_size = 20) +
#  theme_gray(base_size = 20) +
  scale_x_continuous(breaks = seq(from = 0, to = 25, by = 1)) +
  scale_y_continuous(breaks = seq(from = 0, to = 80, by = 10)) +
  coord_cartesian(
    xlim = c(0, 25),
    ylim = c(0, 60)
  )
print(scatterplot3)
```
